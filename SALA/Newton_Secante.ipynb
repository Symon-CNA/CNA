```python
import numpy as np

#PARAMETROS DO PROBLEMA
A = 0.01        # Área (m²)
h = 10          # Coeficiente de convecção (W/m²K)
epsi = 0.8      # Emissividade
T_inf = 300    # Temperatura ambiente (K)
T0_init = 400
T1_init = 425
P = 20          # Potência fornecida (W)
Sig = 5.67e-8   # Constante de Stefan-Boltzmann (W/m²K⁴)

# Função do resíduo do balanço térmico: f(T) = 0
def f(T):
    return h*A*(T - T_inf) + epsi*Sig*A*(T**4 - T_inf**4) - P

# Derivada analítica f'(T)
def df(T):
    return h*A + 4*epsi*Sig*A*(T**3)

# Configurações globais de parada
tol = 1e-6
max_iter = 100

print("="*60)
print("PARTE 1: MÉTODO DE NEWTON-RAPHSON")
print("="*60)

T_k = 350 # Chute inicial

print(f"{'k' :<5} | {'T_k (K)' :<12} | {'f(T_k)' :<15} | {'Erro' :<15}")
print("-"*55)

for k in range(max_iter):
  f_val = f(T_k)
  df_val = df(T_k)

  T_next = T_k - f_val/df_val
  erro = abs(T_next - T_k)

  print(f"{k :<5} | {T_k :<12.6f} | {f_val :<15.6f} | {erro :<15.6e}")

  if erro < tol:
    print(f"\nConvergência alcançada na iteração {k+1}!")
    print(f"Temperatura de equilíbrio T = {T_next:.6f} K\n")
    break
  T_k = T_next


print("="*60)
print("PARTE 2: MÉTODO DA SECANTE")
print("="*60)

# Reiniciando valores para a Secante
T0 = T0_init
T1 = T1_init
iter_sec = 0

print(f"{'k':<5} | {'T_k (K)':<12} | {'f(T_k) (W)':<15} | {'Erro (K)':<15}")
print("-" * 55)

# Apresentação inicial para k = 0
print(f"{0:<5} | {T0_init:<12.6f} | {f(T0_init):<15.6f} | {'-':<15}")

for k in range(1, max_iter):
    f0 = f(T0)
    f1 = f(T1)

    # Fórmula da Secante
    T2 = T1 - f1 * (T1 - T0) / (f1 - f0)
    erro = abs(T2 - T1)

    print(f"{k :<5} | {T1 :<12.6f} | {f1 :<15.6f} | {erro :<15.6e}")

    if erro < tol:
        iter_sec = k
        T_final = T2
        print(f"\nConvegência alcançada na iteração {iter_sec}!")
        print(f"Temperatura de equilíbrio T = {T_final:.6f} K\n")
        break

    # Atualiza os pontos para a próxima iteração
    T0, T1 = T1, T2
```
